## 8. Basic PK/FK validation in Postgres

We perform some simple checks:

- Row counts per table
- FK coverage for key relationships

    
You can extend this with more domain-specific validation, but this gives a structural sanity check that Mostly AI preserved the relations.

````python
with engine.connect() as conn:
    print("Row counts in Postgres:")
    for table_name in [
        "Customer", "Branch", "Account", "Card",
        "Merchant", "Transaction", "Loan", "LoanPayment"
    ]:
        result = conn.execute(text(f"SELECT COUNT(*) FROM \"{table_name}\""))
        count = list(result)[0][0]
        print(f"- {table_name}: {count}")
    
    print("\nFK checks:")
    # Account.customer_id -> Customer.customer_id
    result = conn.execute(text(
        """
        SELECT COUNT(*)
        FROM "Account" a
        LEFT JOIN "Customer" c ON a.customer_id = c.customer_id
        WHERE c.customer_id IS NULL
        """
    ))
    print("Account.customer_id missing Customer:", list(result)[0][0])
    
    # Account.branch_id -> Branch.branch_id
    result = conn.execute(text(
        """
        SELECT COUNT(*)
        FROM "Account" a
        LEFT JOIN "Branch" b ON a.branch_id = b.branch_id
        WHERE b.branch_id IS NULL
        """
    ))
    print("Account.branch_id missing Branch:", list(result)[0][0])
    
    # Card.account_id -> Account.account_id
    result = conn.execute(text(
        """
        SELECT COUNT(*)
        FROM "Card" c
        LEFT JOIN "Account" a ON c.account_id = a.account_id
        WHERE a.account_id IS NULL
        """
    ))
    print("Card.account_id missing Account:", list(result)[0][0])
    
    # Transaction.account_id -> Account.account_id
    result = conn.execute(text(
        """
        SELECT COUNT(*)
        FROM "Transaction" t
        LEFT JOIN "Account" a ON t.account_id = a.account_id
        WHERE a.account_id IS NULL
        """
    ))
    print("Transaction.account_id missing Account:", list(result)[0][0])
    
    # Loan.customer_id -> Customer.customer_id
    result = conn.execute(text(
        """
        SELECT COUNT(*)
        FROM "Loan" l
        LEFT JOIN "Customer" c ON l.customer_id = c.customer_id
        WHERE c.customer_id IS NULL
        """
    ))
    print("Loan.customer_id missing Customer:", list(result)[0][0])
    
    # Loan.account_id -> Account.account_id
    result = conn.execute(text(
        """
        SELECT COUNT(*)
        FROM "Loan" l
        LEFT JOIN "Account" a ON l.account_id = a.account_id
        WHERE a.account_id IS NULL
        """
    ))
    print("Loan.account_id missing Account:", list(result)[0][0])
    
    # LoanPayment.loan_id -> Loan.loan_id
    result = conn.execute(text(
        """
        SELECT COUNT(*)
        FROM "LoanPayment" p
        LEFT JOIN "Loan" l ON p.loan_id = l.loan_id
        WHERE l.loan_id IS NULL
        """
    ))
    print("LoanPayment.loan_id missing Loan:", list(result)[0][0])
````

````python
# 9.a Optional: pyarrow / pandas compatibility shim for Mostly AI demo

import pyarrow as pa

# In some pandas/pyarrow combinations, extension types like "pandas.period"
# can be registered twice, causing ArrowKeyError. This block tries to
# clean up any previous registrations before Mostly AI uses pyarrow.

try:
    from pyarrow import pandas_compat as _pa_pandas_compat
except ImportError:
    _pa_pandas_compat = None

if _pa_pandas_compat is not None:
    try:
        # Best-effort: unregister and re-register extension types.
        # Not all versions expose these helpers, so we guard everything.
        if hasattr(_pa_pandas_compat, "unregister_extension_types"):
            try:
                _pa_pandas_compat.unregister_extension_types()
            except Exception:
                pass
        if hasattr(_pa_pandas_compat, "register_extension_types"):
            try:
                _pa_pandas_compat.register_extension_types()
            except Exception:
                pass
    except Exception as _e:
        print("[INFO] pyarrow/pandas compat shim encountered an issue, but it is safe to ignore:", repr(_e))

print("pyarrow/pandas compatibility shim executed.")
````

````python
# 10.a Initialize Mostly AI client from .env and train on census data

import pandas as pd

from mostlyai_key import get_mostly_client


CENSUS_URL = "https://github.com/mostly-ai/public-demo-data/raw/dev/census/census.csv.gz"


# Load the real census data locally for comparison only
real_df = pd.read_csv(CENSUS_URL)
print("Real census data shape:", real_df.shape)


# Initialize client using API key and base URL from .env
mostly = get_mostly_client()
print("Mostly AI client initialized from .env")


# Train a generator on the public census dataset via URL (per Mostly AI examples)
#
# NOTE:
# - This step may fail if your local pandas/pyarrow versions are
#   incompatible with the Mostly AI SDK (you might see ArrowKeyError
#   or extension type registration errors).
# - To keep the notebook runnable end‑to‑end, we catch any exception
#   and fall back to skipping the synthetic census demo.

synth_df = None

try:
    g = mostly.train(data=CENSUS_URL)
    print("Training finished. Generator ID:", getattr(g, "id", g))

    # Generate a synthetic census dataset
    sd = mostly.generate(g, size=2000)
    synth_df = sd.data()
    print("Synthetic census data shape:", synth_df.shape)
except Exception as e:
    print("[WARN] Mostly AI census demo failed. This is often due to a pandas/pyarrow version mismatch.")
    print("[WARN] Error:", repr(e))
    print("[INFO] Skipping synthetic census generation; downstream checks will detect this and skip.")
````

````python
# 10.b Simple accuracy checks: real vs synthetic census data

import numpy as np



def compare_numeric_stats(real: pd.DataFrame, synth: pd.DataFrame, top_n: int = 5):
    num_cols = [c for c in real.select_dtypes(include=["number"]).columns if c in synth.columns]
    if not num_cols:
        print("No overlapping numeric columns to compare.")
        return
    print("Numeric column mean comparison (first", top_n, "columns):")
    for col in num_cols[:top_n]:
        r_mean = real[col].dropna().mean()
        s_mean = synth[col].dropna().mean()
        print(f"- {col}: real={r_mean:.4f}, synth={s_mean:.4f}, diff={abs(r_mean - s_mean):.4f}")



def compare_categorical_distribution(real: pd.DataFrame, synth: pd.DataFrame, top_n: int = 3):
    cat_cols = [
        c for c in real.select_dtypes(exclude=["number"]).columns
        if c in synth.columns
    ]
    if not cat_cols:
        print("No overlapping categorical columns to compare.")
        return
    col = cat_cols[0]
    print(f"\nCategorical distribution comparison for column: {col}")
    real_counts = (real[col].value_counts(normalize=True).head(top_n) * 100).round(2)
    synth_counts = (synth[col].value_counts(normalize=True).head(top_n) * 100).round(2)
    print("Top real categories (%):")
    print(real_counts)
    print("\nTop synthetic categories (%):")
    print(synth_counts)


# Only run comparisons if synthetic census data was successfully created
if "synth_df" not in globals() or synth_df is None:
    print("[INFO] No synthetic census data available (previous cell failed); skipping accuracy checks.")
else:
    compare_numeric_stats(real_df, synth_df)
    compare_categorical_distribution(real_df, synth_df)
````
